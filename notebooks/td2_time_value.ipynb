## Part 4. IRR (20 min)

The IRR is the discount rate at which the NPV equals zero. No formula gives it: you try several rates and close in on the zero. The function below gives the NPV of a constant cash flow at any rate. Run it.

```python
def npv_constant(cash_flow, rate, years=20):
    """NPV of a constant yearly cash flow, CAPEX paid in year 0."""
    annuity = (1 - (1 + rate) ** -years) / rate
    return -CAPEX + cash_flow * annuity
```
